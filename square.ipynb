```python
from jetbot import Robot
```

```python
robot = Robot()
```

```python
import time
```

```python
def step_forward():
    robot.forward(0.3)
    time.sleep(1)
    robot.stop()
    
def step_right():
    robot.right(0.4)
    time.sleep(0.5)
    robot.stop()
    
step_forward()
step_right()
step_forward()
step_right()
step_forward()
step_right()
step_forward()
step_right()
```

```python
robot.stop()
```
