```python
from jetbot import Robot
```

```python
robot = Robot()
```

```python
import time
```

```python
robot.stop()
```

```python
def stop():
    robot.stop()
    
def step_forward():
    robot.forward(0.256)
    time.sleep(1.0)
    robot.stop()

def step_right():
    robot.right(0.197)
    time.sleep(0.)
    robot.stop()
    
for i in range(4):
  step_forward()    
  step_right()
  stop()
```
